```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
```

```python
business = pd.read_json("dataset/yelp_academic_dataset_business.json", lines=True)
```

```python
business.shape
```

```python
checkin = pd.read_json("dataset/yelp_academic_dataset_checkin.json", lines=True)
```

```python
checkin.shape
```

```python
tips = pd.read_json("dataset/yelp_academic_dataset_tip.json", lines=True)
```

```python
tips.shape
```

```python
chunks = []

for chunk in pd.read_json("dataset/yelp_academic_dataset_review.json", lines=True, chunksize=100000):
    chunks.append(chunk)

reviews = pd.concat(chunks, ignore_index=True)
```

```python
reviews.shape
```

```python
business = business.drop(columns=["attributes", "hours"])
```

```python
conn = sqlite3.connect('yelp.db')

business.to_sql('business', con=conn, if_exists='replace', index=False)
checkin.to_sql('checkin', con=conn, if_exists='replace', index=False)
tips.to_sql('tips', con=conn, if_exists='replace', index=False)
reviews.to_sql('reviews', con=conn, if_exists='replace', index=False)

conn.close()
```
